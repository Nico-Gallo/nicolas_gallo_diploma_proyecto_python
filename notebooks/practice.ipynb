En el presente proyecto, el objetivo es poder analizar datos de Egresos Hospitalarios en Uruguay entre los años 2021 y 2024.
El dataset fue obtenido del Catálogo de Datos Abierto y se seleccionaron estos años para el análisis por ser los datos disponibles más actuales a la fecha.

En el dataset, se encuentran datos de los egresos hospitalarios en Uruguay, significando esto casos en los que la persona abandonó el centro hospitalario pudiendo ser consecuencia de su fallecimiento o de su alta domiciliaria, sin er ello explícitamente identificado. Dentro de los datos, se encuentra no solamente el año, sino también el género, región, grupo etareo, sistema de salud, lo que permite cruzarlos para lograr obtener información relevante sobre tendencias.

Comenzando con la fase 1 del trabajo, importo las librerías a utilizar y traigo el dataset intervenido para proceder a su EDA.
Presento situación primaria del dataset y contabilizo los valores nulos que serán limpiados.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

base_dir = os.path.abspath(os.path.join(os.getcwd(), '..'))
ruta = os.path.join(base_dir, 'data', 'raw', 'p4ds_grupo_c_egresos_hospitalarios.csv')

print("Buscando en:", ruta)

df = pd.read_csv(ruta, sep=';', dtype_backend='numpy_nullable')

display(df.head())
print(f"Filas: {df.shape[0]:,} | Columnas: {df.shape[1]}")
df.info()
print("\n- VALORES NULOS -")
print(df.isnull().sum())
```

Presento los valores de cada columna que son únicos, para identificar también posibilidades futuras de visualización.

```python
print("- VALORES ÚNICOS POR COLUMNA -")
for col in df.columns:
    print(f"\n{col} ({df[col].nunique()} únicos):")
    print(df[col].unique()[:10])  # primeros 10 valores únicos
```

Identifico filas que se encuentran dobles en el dataset para poder limpiarlas y se muestran algunas de estas filas.

```python
print("- DUPLICADOS -")
print(f"Filas duplicadas: {df.duplicated().sum():,}")
print(f"IDs duplicados: {df['ID'].duplicated().sum():,}")
```

```python
duplicados = df[df.duplicated(keep=False)]
print(f"Total de filas involucradas en duplicados: {len(duplicados):,}")
display(duplicados.sort_values('ID').head(10))
```

Elimino las duplicadas, manteniendo aquella que aparece primero y presento cuantificación de datos originales y posteriores.

```python
df_limpio = df.drop_duplicates(keep='first')

print(f"Filas originales:  {len(df):,}")
print(f"Filas después:     {len(df_limpio):,}")
print(f"Filas eliminadas:  {len(df) - len(df_limpio):,}")
```

En algunas columnas se presentas algunos datos que debo evaluar que hacer para limpiarlos: unificar o eliminar. Los presento y cuantifico.

```python
print("- VALORES A LIMPIAR -")

print(f"\nREGION - nulos: {df_limpio['REGION'].isna().sum():,}")
print(f"REGION - 'Sin datos': {(df_limpio['REGION'] == 'Sin datos').sum():,}")
print(f"REGION - 'EXTERIOR': {(df_limpio['REGION'] == 'EXTERIOR').sum():,}")

print(f"\nGRUPO ETAREO - 'Sin datos': {(df_limpio['GRUPO ETAREO'] == 'Sin datos').sum():,}")

print(f"\nCAUSA EXTERNA - '99': {(df_limpio['CAUSA EXTERNA'] == '99').sum():,}")
print(f"CAUSA EXTERNA - 'SD': {(df_limpio['CAUSA EXTERNA'] == 'SD').sum():,}")
print(f"CAUSA EXTERNA - 'NC': {(df_limpio['CAUSA EXTERNA'] == 'NC').sum():,}")
```

De la anterior presentación y cuantificación identifico lo siguiente para cada columna:

REGION: unifico "nulos" y "sin datos" en "sin datos", que para el caso son lo mismo.

GRUPO ETAREO: la cantidad de "sin datos" es totalmente insignificante por lo que no hago cambios.

CAUSA EXTERNA: aquellos datos presentados como "SD" y "99" los unifico en "Sin datos" ya que a estos efectos significan lo mismo. Y los presentados como "NC", que significan "no corresponde", es decir que se entiende que no hubo una causa externa, son totalmente válidos por lo que los mantengo tal como estan.

Luego presento y cuantifico los datos de estas columnas.

```python
# Elimino registros de EXTERIOR
df_limpio = df_limpio[df_limpio['REGION'] != 'EXTERIOR']
print(f"Filas después de eliminar EXTERIOR: {len(df_limpio):,}")

# Unifico nulos y 'Sin datos' en 'Sin datos'
df_limpio['REGION'] = df_limpio['REGION'].fillna('Sin datos')

# Unifico '99' y 'SD' en 'Sin datos'
df_limpio['CAUSA EXTERNA'] = df_limpio['CAUSA EXTERNA'].replace({'99': 'Sin datos', 'SD': 'Sin datos'})

print("- VERIFICACIÓN POST LIMPIEZA -")
print("\nREGION:")
print(df_limpio['REGION'].value_counts())
print("\nCAUSA EXTERNA:")
print(df_limpio['CAUSA EXTERNA'].value_counts())
```

Entendiendo que el dataset original fue depurado, presento un resumen del final, antes de grabarlo en la sección solicitada.

```python
print("- RESUMEN FINAL DEL DATASET LIMPIO -")
print(f"\nFilas originales:  1,132,011")
print(f"Filas finales:     {len(df_limpio):,}")
print(f"Columnas:          {df_limpio.shape[1]}")
print(f"\nValores nulos restantes:")
print(df_limpio.isnull().sum())
```

Grabo el CSV limpio en data/processed/:

```python
ruta_salida = os.path.join(base_dir, 'data', 'processed', 'egresos_hospitalarios_limpio.csv')

df_limpio.to_csv(ruta_salida, index=False, sep=';')

print(f"Archivo grabado en: {ruta_salida}")
print(f"Filas grabadas: {len(df_limpio):,}")
```

- Unificación de '99' y 'SD' en CAUSA EXTERNA
